```python
from pathlib import Path

PROJECT_DIR = Path(
    "/content/drive/MyDrive/traffic-accident-risk-analytics"
)

readme = """# AI-Powered Traffic Congestion and Road Accident Risk Prediction & Analytics System

## Project Overview

This project investigates historical traffic accident patterns and the reliability of machine-learning models used for road-crash severity prediction.

The study will focus on publicly available crash records for Dallas and Tarrant Counties, Texas, subject to data availability. In addition to exploratory data analysis and geographic hotspot analysis, the project will examine whether model performance changes across different years and locations.

The research will also evaluate probability calibration and investigate whether a model can improve practical reliability by abstaining from predictions on uncertain cases.

The final deliverable will be an interactive analytics dashboard that communicates historical crash patterns, model performance, prediction uncertainty, and limitations.


## Problem Statement


Machine-learning models can perform well on a test dataset but may become less reliable when applied to data from later years or different geographic locations.

This project aims to investigate the performance and reliability of crash-severity classification models using available historical Texas crash records. It will compare model performance under different evaluation conditions, assess probability calibration, and explore selective prediction, where a model abstains from predicting cases identified as uncertain.

The objective is to build a reproducible research pipeline and interactive decision-support dashboard that communicates not only predictions but also their limitations.


## Objectives

1. Collect and inspect publicly available historical crash records.
2. Clean the data and evaluate missing values, duplicates, and class imbalance.
3. Explore crash patterns by time, severity, weather, roadway conditions, and location where available.
4. Identify historical crash hotspots.
5. Develop and compare baseline and advanced crash-severity classification models.
6. Evaluate model performance across different years and, where feasible, geographic areas.
7. Assess probability calibration and investigate uncertainty-aware selective prediction.
8. Explain model outputs using feature importance and suitable interpretation methods.
9. Build an interactive dashboard to present analytical results and model reliability.


## Research Questions

**RQ1:** How does crash-severity classification performance change when testing on later years instead of a random test sample?

**RQ2:** How does model performance change when tested on a geographic area not represented in training, where the dataset supports this comparison?

**RQ3:** How well calibrated are the predicted probabilities under different evaluation conditions?

**RQ4:** Can abstaining on uncertain cases reduce prediction errors, and what proportion of cases can still receive predictions at different error levels?

**RQ5:** Which available temporal, environmental, roadway, and geographic features contribute most to model predictions?


## Initial Scope


- **Geography:** Dallas and Tarrant Counties, Texas.
- **Target period:** January 1, 2021, through December 31, 2025, subject to data availability.
- **Primary data:** Public crash-level records from the Texas Department of Transportation.
- **Primary prediction task:** Crash-severity classification, subject to the actual dataset fields and definitions.
- **Additional analyses:** Historical crash hotspots, temporal patterns, model reliability, and uncertainty-aware decision support.


## Planned Technology

Python, Pandas, NumPy, Matplotlib, Seaborn, Scikit-learn,
Google Colab, Google Drive, Git, GitHub, and Streamlit.

## Project Status

Phase 1: Project setup and research.

## Data and Limitations

Data sources, cleaning procedures, prediction targets, model performance,
and limitations will be documented as the project progresses.
Historical associations do not by themselves establish causation.
"""

requirements = """pandas
numpy
matplotlib
seaborn
scikit-learn
streamlit
plotly
folium
shap
"""

gitignore = """# Python
__pycache__/
*.py[cod]
.venv/
venv/

# Jupyter
.ipynb_checkpoints/

# Environment and credentials
.env
*.key
credentials.json
token.json

# Local data and model artifacts
data/raw/*
data/processed/*
data/external/*
models/*
!data/raw/.gitkeep
!data/processed/.gitkeep
!data/external/.gitkeep
!models/.gitkeep

# Local system files
.DS_Store
"""

files = {
    "README.md": readme,
    "requirements.txt": requirements,
    ".gitignore": gitignore,
}

for filename, content in files.items():
    path = PROJECT_DIR / filename
    path.write_text(content, encoding="utf-8")
    print("Created:", path)
```
