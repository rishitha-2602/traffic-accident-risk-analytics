```python
from pathlib import Path

PROJECT_DIR = Path(
    "/content/drive/MyDrive/traffic-accident-risk-analytics"
)

readme = """# AI-Powered Traffic Congestion and Road Accident Risk Prediction & Analytics System

## Project Overview

This Data Science capstone investigates historical road accident patterns
and develops an analytics system for studying crash hotspots and suitable
machine-learning prediction tasks in the Dallas–Fort Worth area of Texas.

## Problem Statement

Road accidents can be associated with temporal, environmental, roadway,
and traffic-related factors. This project analyzes historical crash records
to understand these patterns and evaluate data-driven prediction methods.

## Objectives

1. Collect and inspect suitable historical crash data.
2. Clean and prepare the data for analysis.
3. Explore accident patterns by time, severity, weather, and location.
4. Identify geographic concentrations of reported crashes.
5. Train and compare suitable machine-learning models.
6. Interpret model outputs and communicate findings.
7. Develop an interactive dashboard.

## Research Questions

- RQ1: Which temporal, environmental, and roadway factors are associated
  with recorded crashes?
- RQ2: Which locations show the highest historical crash concentrations?
- RQ3: How well do selected models predict the chosen outcome?
- RQ4: Which features contribute most to model predictions?

## Initial Scope

- Study area: Dallas and Tarrant Counties, Texas.
- Historical period: 2021–2025, subject to data availability.
- Primary data: Historical crash records.
- Additional data: Compatible traffic, weather, and geographic data
  where available.

## Planned Technology

Python, Pandas, NumPy, Matplotlib, Seaborn, Scikit-learn,
Google Colab, Google Drive, Git, GitHub, and Streamlit.

## Project Status

Phase 1: Project setup and research.

## Data and Limitations

Data sources, cleaning procedures, prediction targets, model performance,
and limitations will be documented as the project progresses.
Historical associations do not by themselves establish causation.
"""

requirements = """pandas
numpy
matplotlib
seaborn
scikit-learn
streamlit
plotly
folium
shap
"""

gitignore = """# Python
__pycache__/
*.py[cod]
.venv/
venv/

# Jupyter
.ipynb_checkpoints/

# Environment and credentials
.env
*.key
credentials.json
token.json

# Local data and model artifacts
data/raw/*
data/processed/*
data/external/*
models/*
!data/raw/.gitkeep
!data/processed/.gitkeep
!data/external/.gitkeep
!models/.gitkeep

# Local system files
.DS_Store
"""

files = {
    "README.md": readme,
    "requirements.txt": requirements,
    ".gitignore": gitignore,
}

for filename, content in files.items():
    path = PROJECT_DIR / filename
    path.write_text(content, encoding="utf-8")
    print("Created:", path)
```
